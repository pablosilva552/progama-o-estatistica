```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,4, 100000)
y = np.linspace(-4,4, 100000)
```

```python
def normal(x):
    return 2*(1/(np.pi*2)**0.5 * np.exp(-x**2/2))
```

```python
def g(y):
    lamb = 1
    return lamb * np.exp(-lamb * y)
```

```python
c = 2*normal(1)/g(1)
print(c)
```

```python
n = 100_000
amostra = []
sucessos = 0
tentativa = 0 
while sucessos < n:
    u = np.random.uniform()
    ub = -np.log(np.random.uniform())
    if u*c <= normal(ub)/g(ub):
        amostra.append(ub)
        sucessos += 1
plt.hist(amostra, density = True, bins = 200)
plt.plot(y, normal(y))

```

```python
p = 0.5 
amostra_2 = []
for x in amostra:
    u = np.random.uniform()
    if u < p:
        amostra_2.append(x)
    else:
        amostra_2.append(-x)
```

```python
plt.hist(amostra_2, density = True, bins = 250)
plt.plot(y, normal(y)/2)
```

```python
plt.hist(np.random.normal(0,1, 100_000), density = True, bins = 250)
```

```python
def normal_mu_sigma2(mu, sigma2):
    return mu + sigma2 * amostra_2
```

```python
amostra_2 = np.array(amostra_2)
plt.hist(normal_mu_sigma2(32,22), density = True, bins = 25)
```
