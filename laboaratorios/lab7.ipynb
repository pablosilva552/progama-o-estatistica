```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,1,10000)
```

```python
def funcao(x):
    f = (20 * x)*(1-x)**3
    return f
```

```python
n = 10_000
amostra = []
tentativas = 0
sucesso = 0
while sucesso < n:
    y = np.random.uniform(0,1)
    U = np.random.uniform(0,1)
    c = 2.1
    if U < funcao(y)/c:
        amostra.append(y)
        sucesso += 1
    tentativas += 1
plt.hist(amostra, density=True)
plt.plot(x, funcao(x))
```

```python
np.max(funcao(x))
```

```python
v =tentativas / sucesso
print(v)
```
