```python
def gama(y, alpha, lam):
    return (lam**alpha/math.gamma(alpha)) * (y**(alpha - 1)) * np.exp(lam - 1)  
```

```python
def exp(x, lam):
    return lam*np.exp(-lam*x)

def gama(x, alpha, lam):
    return (lam**alpha/math.gamma(alpha)) * (x**(alpha - 1)) * np.exp(-lam*x)

n = 100_000
a = 2.5
lam = 2
mi = lam/a                     
x_est = (a - 1)/(lam - mi)
c = gama(x_est, a, lam)/exp(x_est, mi)

amostra = []
sucesso = 0
tentativa = 0
while sucesso < n:
    ub = np.random.uniform()
    y = -np.log(np.random.uniform())/mi
    if ub * c * exp(y, mi) <= gama(y, a, lam):
        amostra.append(y)
        sucesso += 1
    tentativa += 1

print(sucesso/tentativa, 1/c)
```

```python
plt.hist(amostra, density = True, bins = 50 )
```

## exercicio 4

```python
def beta(x, alpha, bet):
    B = math.gamma(alpha)*math.gamma(bet)/math.gamma(alpha + bet)
    return (x**(alpha - 1)) * ((1 - x)**(bet - 1)) / B
```

```python
n = 100_000
a = 2.5
b = 4
x_est = (a - 1)/(a + b - 2)
c = beta(x_est, a, b) 
amostra = []
sucesso = 0
tentativa = 0
while sucesso < n:
    ub = np.random.uniform()
    y = np.random.uniform()
    if ub * c <= beta(y, a, b):
        amostra.append(y)
        sucesso += 1
    tentativa += 1

print(sucesso/tentativa, 1/c) 
```

```python
amostra_numpy = np.random.beta(a, b, size=n)

plt.hist(amostra, density=True, bins=50, label='Aceitação-rejeição')
plt.hist(amostra_numpy, density=True, bins=50, label='NumPy')
plt.legend()
plt.show()
```
