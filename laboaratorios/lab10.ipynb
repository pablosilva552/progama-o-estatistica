## exercicio 1

```python
import numpy as np

n = 100_000
z = 1.96
```

```python
x = np.random.uniform(size=n)
g = np.exp(-x**2)

I_est = np.mean(g)
var_est = np.var(g, ddof=1)/n   
ic = (I_est - z*np.sqrt(var_est), I_est + z*np.sqrt(var_est))

print(f'{I_est}, {var_est}, {ic}')
```

## Exercicio 2

```python
x = np.random.normal(0, 1, size=n)
ind = (x > 1)

p_est = np.mean(ind)
var_est = p_est*(1 - p_est)/n
ic = (p_est - z*np.sqrt(var_est), p_est + z*np.sqrt(var_est))

print(p_est, var_est, ic)
```

## Exercicio 3

```python
x = np.random.uniform(size=n)
y = np.random.uniform(size=n)
ind = (x**2 + y**2 <= 1)

p_est = np.mean(ind)
pi_est = 4*p_est
var_est = 16*p_est*(1 - p_est)/n
ic = (pi_est - z*np.sqrt(var_est), pi_est + z*np.sqrt(var_est))

print(pi_est, var_est, ic)
```

## Exercicio 4


```python
x = np.random.uniform(size=n)

media_est = np.mean(x)
var_x_est = np.var(x, ddof=1)  
var_media = var_x_est/n                  
ic = (media_est - z*np.sqrt(var_media), media_est + z*np.sqrt(var_media))

print(media_est, var_x_est, var_media, ic)
```
