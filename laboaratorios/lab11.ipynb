```python

import matplotlib.pyplot as plt
import numpy as np
import scipy.special as special
```

```python
n = 100_000
lam = 2
gamma = []
for i in range(n):
  exp = []
  for i in range(3):
    u = np.random.uniform()
    y = -np.log(u)/lam
    exp.append(y)
  gamma.append(np.sum(exp))  #X~Gamma(3,2)

amostra = []
for j in gamma:
  amostra.append(j**0.5)  #raiz de X, X~Gamma(3,2)

media = np.mean(amostra)
var = np.var(amostra)/n  #variância do estimador
ic = (media - 2*np.sqrt(var),media + 2*np.sqrt(var))
valor_teorico = 1.174982
print(f'O valor teórico {valor_teorico}, está contido no intervalo de confiança {ic}')
```

```python

def f(x):
  return (x**2 + 2*x + 1)
amostra_2 = []
for i in range(n):
  u = np.random.uniform()*2*np.pi
  u2 = np.random.uniform()
  t = np.sqrt(-2*np.log(u2))*np.cos(u)
  x = 1 + 2*t
  amostra_2.append(x)

integral = []
for j in amostra_2:
  integral.append(f(j))

media = np.mean(integral)
var = np.var(integral)/n
ic = (media - 2*np.sqrt(var), media + 2*np.sqrt(var))
print(f'O valor teórico da probabilidade, I=8, está contido no intervalo de confiança {ic}')
     
```

## Exercicio 5 

```python
n = 100_000
z = 1.96

c = 2.0736
amostras = np.zeros(n)
aceitos = 0

while aceitos < n:
    faltam = n - aceitos
    
    y = np.random.uniform(0, 1, faltam)
    
    u2 = np.random.uniform(0, 1, faltam)
    
    f_y = 60 * (y**2) * ((1 - y)**3)
    g_y = 1
    
    condicao = u2 <= (f_y / c)
    
    y_aceitos = y[condicao]
    qtd = len(y_aceitos)
    
    if qtd > 0:
        limite = min(aceitos + qtd, n)
        amostras[aceitos:limite] = y_aceitos[:(limite - aceitos)]
        aceitos = limite

x = amostras

y_val = np.exp(x) / 60

I_est = np.mean(y_val)

var = np.var(y_val, ddof=1) / n

erro = z * np.sqrt(var)
ic = [I_est - erro, I_est + erro]

valor_teorico = 0.02598262

print(f"I_est: {I_est:.8f}")
print(f"Variância: {var:.10f}")
print(f"IC 95%: [{ic[0]:.8f}, {ic[1]:.8f}]")
print(f"Valor Teórico: {valor_teorico}")
print(f"Erro absoluto: {abs(I_est - valor_teorico):.8f}")
```

```python

normal_1 = []
for i in range(n):
  u = np.random.uniform()*2*np.pi
  u2 = np.random.uniform()
  t = np.sqrt(-2*np.log(u2))*np.cos(u) 
  normal_1.append(t)

normal_2 = []
for j in range(n):
  u = np.random.uniform()*2*np.pi
  u2 = np.random.uniform()
  v = np.sqrt(-2*np.log(u2))*np.sin(u)
  normal_2.append(v)

indicadora = []
for k in range(n):
    if normal_1[k]**2 + normal_2[k]**2 > 5:
      indicadora.append(1)
    else:
      indicadora.append(0)

media = np.mean(indicadora)
var = np.var(indicadora)/n
ic = (media - 2*np.sqrt(var), media + np.sqrt(var))

print(f'Valor teórico: 0.082085, Valor do estimador: {media}, o ic é {ic}')
```

```python
x1 = []
x2 = []
x3 = []
x4 = []
x5 = []

for i in range(n):
  u1 = np.random.uniform()
  u2 = np.random.uniform()
  u3 = np.random.uniform()
  u4 = np.random.uniform()
  u5 = np.random.uniform()
  y1 = -np.log(u1)/2
  y2 = -np.log(u2)/2
  y3 = -np.log(u3)/2
  y4 = -np.log(u4)/2
  y5 = -np.log(u5)/2
  x1.append(y1), x2.append(y2), x3.append(y3), x4.append(y4), x5.append(y5)

maximo = []
for j in range(n):
  maximo.append(max(x1[j], x2[j], x3[j], x4[j], x5[j]))

media = np.mean(maximo)
var = np.var(maximo)/n
ic = (media - 2*np.sqrt(var), media + 2*np.sqrt(var))

print(f'E[M] teórica é 1.141667 e o valor estimado por monte carlo é {media}, o ic é {ic}')
     
```
