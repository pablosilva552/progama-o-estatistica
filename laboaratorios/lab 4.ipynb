```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
N = 10000
U = np.random.uniform(size = N)
print(U)
```

aqui foi criado uma amostra aleatoria de uniformes que vai ser usado pra criar as variaveis aleatorias seguinte


```python
X = []
for i in range(N):
        if U[i] < 0.5:
                X.append(1)
        else:
                X.append(0)
print(X)
```

Esse código basicamente define uma lista vazia X,e cria um for que faz as comparações de cada U e adiciona a X como 0 ou 1 criando uma amostra de bernoulli

```python
y = 1
for u in U:
    if u < 0.5:
        break
    y += 1

print(y)
```

Nesse aqui o conceito é similar, mas aqui estamos somando sucessos e adicionando criando uma binomial

```python
z = []  
y = 1   
i = 0   
while i < len(U):
    if U[i] < 0.5:
        z.append(y)
        y = 1
    else:
        y += 1
    
    i += 1
print(z)
```

como foi pedido aqui criamos uma amostra de binomias

```python
plt.hist(z)
```

Foi plotado o gráfico de z que é uma amostra de binomias

```python
v = np.random.geometric(0.5,size=10000)
```

Usando a biblioteca numpy criei uma amostra de geometricas com os mesmos parametros que usei pra simular por as uniformes

```python
plt.hist(v)
```

### Pela inversão 


```python
p = 0.5
m = np.random.uniform(0,1,N)
j = np.ceil(np.log(1-m)/np.log(1-p) + 1)
```

Pela inversão e usando a formula fechada simulei uma geometrica

```python
plt.hist(j)
```

```python
k = np.arange(1, 20)

prob_j = np.array([np.mean(j == i) for i in k])
prob_v = np.array([np.mean(v == i) for i in k])

teorico = 0.5 * (1 - 0.5)**(k - 1)

largura = 0.25

plt.bar(k - 2*largura, prob_j, width=largura, label='Inversão')
plt.bar(k, prob_v, width=largura, label='NumPy')
plt.bar(k + largura, teorico, width=largura, label='Teórico')

plt.xlabel('k')
plt.ylabel('Probabilidade')
plt.title('Distribuição Geométrica (p=0.5)')
plt.xticks(k)
plt.legend()

plt.show()
```

### Exercício 3 poisson

```python
N = 1000
U = np.random.uniform(size = N)
z = []
lamb = 3
for u in U :
    i = 0 
    p = np.exp(-lamb)
    F = p
    while u > F:
        i += 1
        p = (lamb/i) * p    
        F += p
    z.append(i)
print(z)    
```

aqui eu vou criar uma amostra de poisson, criando uma amostra aleatoria uniforme, e usando formula de inversão.

```python

plt.hist(z, density=True)
plt.xlabel('k')
plt.ylabel('Probabilidade')
plt.show()
```

plotei o grafico de z e logo abaixo o de a para comparar se estão parecidas a que eu criei e a teorica e saber se o método que eu usei está correto 

```python
a = np.random.poisson(lam=lamb, size=1000)

plt.hist(a, density=True)
plt.xlabel('k')
plt.ylabel('Probabilidade')
plt.show()
```
